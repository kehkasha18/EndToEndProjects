# **Parts of Speech (POS) Tagging with NLTK**

### 1. Introduction and Importance

*   **Definition:** Parts of Speech (POS) Tagging involves automatically categorizing words in a sentence into their corresponding parts of speech (e.g., noun, verb, adjective).
*   **Role in NLP:** POS tagging is **very important** in Natural Language Processing (NLP), especially for processes like **Lemmatization**.
*   **Lemmatization Link:** POS tagging helps determine the root form of a specific word (the lemma) by identifying whether the word is a verb or a noun.
*   **Core Tool:** The lecture uses the **NLTK** (Natural Language Toolkit) library to perform POS tagging.

**Parts of Speech (POS) Tagging** is the process of assigning a grammatical category (like noun, verb, adjective, pronoun etc.) to each word in a sentence.

Example sentence:
`Taj Mahal is a beautiful monument`

POS output (example):

* Taj → Noun (Proper Noun)
* Mahal → Noun (Proper Noun)
* is → Verb
* a → Determiner
* beautiful → Adjective
* monument → Noun

> POS tagging helps the computer understand grammar and structure of language.

### 🧠 What is **`punkt`**?

**`punkt`** is a **pre-trained tokenizer model** that comes with **NLTK** (Natural Language Toolkit).
It’s used to **split text into sentences or words** — this process is called **tokenization**.

When you write:

```python
nltk.download('punkt')
```

you’re **downloading the data files** that NLTK needs to perform sentence and word tokenization.
Without downloading `punkt`, NLTK wouldn’t know how to identify where one sentence ends and another begins.


````python
import nltk
nltk.download('stopwords')
````

* Stopwords are not a package — they are a dataset/resource that NLTK uses.
* It is a dataset / corpus that contains lists of commonly used words in many languages like English, French, German, etc.

* NLTK gives you access to this corpus, but you must download it once before using.

# **Averaged Perceptron Tagger**

It is an **algorithm/model** that assigns grammatical labels to words in a sentence — like **noun, verb, adjective, pronoun**, etc.

Example:

```python
import nltk
nltk.download('averaged_perceptron_tagger')
nltk.pos_tag(["Taj", "Mahal", "is", "beautiful"])
```

Output:

```
[('Taj', 'NNP'), ('Mahal', 'NNP'), ('is', 'VBZ'), ('beautiful', 'JJ')]
```

| Tag | Meaning                            |
| --- | ---------------------------------- |
| NNP | Proper noun                        |
| VBZ | Verb (3rd person singular present) |
| JJ  | Adjective                          |

### **Why do we need to download it?**

NLTK only comes with the **code**.
This tagger is a **language model** trained on large text data, so it must be downloaded separately.

### 🍔 Think of it like this:

| Component | Example                               |
| --------- | ------------------------------------- |
| Library   | `nltk` (the kitchen)                  |
| Resource  | Stopwords list (ingredients)          |
| Model     | Averaged perceptron tagger (the chef) |

You're downloading the *chef* that knows how to label each word grammatically.

### **Why is it called Averaged Perceptron?**

Because it uses a **Perceptron algorithm** (a simple ML model) and averages weights over iterations to improve accuracy.

You don’t need to train anything — NLTK gives you a pre-trained model.


###  In simple words

> `averaged_perceptron_tagger` = A **pretrained NLP model** that helps Python understand grammar of each word.

# Line-by-line explanation (simple)

Nice — let’s walk through this together, one line at a time, in plain English.

```python
## We will find the Pos Tag

for i in range(len(sentences)):
    words = nltk.word_tokenize(sentences[i])
    words = [word for word in words if word not in set(stopwords.words('english'))]
    # sentences[i] = ' '.join(words)  # converting all the list of words into sentences
    pos_tag = nltk.pos_tag(words)
    print(pos_tag)
```

1. `for i in range(len(sentences)):`

   * Loop through every sentence in the `sentences` list.
   * `i` is the index (0, 1, 2, …).
   * In simple terms: “Do the following steps for each sentence.”

2. `words = nltk.word_tokenize(sentences[i])`

   * Take the `i`-th sentence (a string) and split it into a list of words and punctuation using NLTK’s tokenizer.
   * Example: `"I love NLP!"` → `['I', 'love', 'NLP', '!']`.
   * Note: `word_tokenize` uses the `punkt` model, so you may need `nltk.download('punkt')` once.

3. `words = [word for word in words if word not in set(stopwords.words('english'))]`

   * This is a list comprehension that **keeps only words that are not stopwords**.
   * `stopwords.words('english')` returns a list like `['the', 'is', 'and', ...]`. `set(...)` makes membership checks faster.
   * Result: `['love', 'NLP']` from the earlier example (removes `'I'` if `'I'` is in stopwords).
   * In plain words: “Remove small common words that often don’t add much meaning.”

   **Small improvement note:** calling `set(stopwords.words('english'))` inside the loop recomputes it every sentence — better to compute it once before the loop:

   ```python
   stop_words = set(stopwords.words('english'))
   ...
   words = [w for w in words if w not in stop_words]
   ```

4. `# sentences[i] = ' '.join(words)  # converting all the list of words into sentences`

   * This line is commented out (ignored). If used, it would join the filtered words back into a single string and replace the original sentence.
   * Example: `['love', 'NLP']` → `"love NLP"`.
   * Right now it’s commented because you want to work with the word list for POS tagging, not a single string.

5. `pos_tag = nltk.pos_tag(words)`

   * Ask NLTK to assign a POS tag to each word in the `words` list.
   * Output is a list of `(word, tag)` pairs. Example: `('love', 'VBP')`, `('NLP', 'NNP')`.
   * Important: This requires the averaged perceptron tagger resource: `nltk.download('averaged_perceptron_tagger')`.

6. `print(pos_tag)`

   * Show the POS tagged list for the current sentence on screen/console.
   * Example printed output for one sentence might look like:

     ```
     [('Taj', 'NNP'), ('Mahal', 'NNP'), ('beautiful', 'JJ'), ('monument', 'NN')]
     ```

## **Tiny optimization suggestion**

### Calculate stopwords `set` once before the loop:

````python
stop_words = set(stopwords.words('english'))
for i in range(len(sentences)):
    words = nltk.word_tokenize(sentences[i])
    words = [w for w in words if w not in stop_words]
    pos_tag = nltk.pos_tag(words)
    print(pos_tag)
````

## 🧠 What is POS Tagging?

POS Tagging = **telling the computer the job of each word** in a sentence.

Examples:

* *Noun* → Person/place/thing
* *Verb* → Action
* *Adjective* → Describes something
* *Adverb* → Describes the action
* Pronoun, Preposition, Determiner… etc.

> "Time flies like an arrow"
